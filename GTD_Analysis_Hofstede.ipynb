```python
import pandas as pd
import matplotlib.pyplot as plt


attacchi_per_paese = df_attack_rel['country_txt'].value_counts().head(10)

# Crea il grafico
plt.figure(figsize=(10, 5))
bars = attacchi_per_paese.plot(kind='bar')

plt.title("Top 10 paesi per numero di attacchi")
plt.xlabel("Paese")
plt.ylabel("Numero di attacchi")
plt.xticks(rotation=45, ha='right')

# Aggiungi i numeri sopra ogni barra
for i, value in enumerate(attacchi_per_paese):
    plt.text(i, value + (value * 0.01), str(value), ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Gruppi terroristici (matrice religiosa)più attivi per paese
group_country = df_attack_rel.groupby(['gname', 'country_txt']).size().reset_index(name='num_attacks')
# Escludi "Unknown" se vuoi solo i gruppi noti
group_country = group_country[group_country['gname'] != 'Unknown']
# Gruppi più attivi nel periodo
print(group_country.sort_values('num_attacks', ascending=False).head(10))
```

```python
import pycountry

def get_iso3(country_name):
    try:
        return pycountry.countries.lookup(country_name).alpha_3
    except:
        return None

# Supponendo che la colonna dei paesi si chiami 'Paese Principale'
group_country['iso_alpha'] = group_country['country_txt'].apply(get_iso3)
```

```python
group_country.to_csv("group_country_iso.csv")
```
