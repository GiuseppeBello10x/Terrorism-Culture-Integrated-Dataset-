```python

# Conta il numero di attacchi per paese
attacks_per_country = df_attack_rel['country_txt'].value_counts().reset_index()
attacks_per_country.columns = ['country', 'num_attacks']
```

```python
import pycountry

def get_iso3(country_name):
    try:
        return pycountry.countries.lookup(country_name).alpha_3
    except:
        return None

attacks_per_country['iso_alpha'] = attacks_per_country['country'].apply(get_iso3)
# Filtra eventuali paesi senza codice ISO
attacks_per_country = attacks_per_country.dropna(subset=['iso_alpha'])
```

```python
import plotly.express as px

fig = px.choropleth(
    attacks_per_country,
    locations="iso_alpha",
    color="num_attacks",
    hover_name="country",
    color_continuous_scale="Reds",
    title="Heatmap degli attacchi terroristici per paese (2018-2021)"
)
fig.show()
```
